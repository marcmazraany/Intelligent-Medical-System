```python
import cv2
from paddleocr import PaddleOCR

ocr = PaddleOCR(lang='en')

img_path = "images/Panadol-Image.jpeg"
img = cv2.imread(img_path)

if img is None:
    raise FileNotFoundError(f"Could not find image: {img_path}")

result = ocr.predict(img)

print("=" * 60)
print("DETECTED TEXT:")
print("=" * 60)

if result and result[0]:
    data = result[0]
    
    if 'rec_texts' in data:
        texts = data['rec_texts']
        scores = data.get('rec_scores', [])
        print(f"\nFound {len(texts)} text regions:\n")
        for i, (text, score) in enumerate(zip(texts, scores), 1):
            print(f"{i}. {text} (confidence: {score:.3f})")
    else:
        print("Recognition data not found in result")
        print(f"Available keys: {data.keys()}")
else:
    print("No text detected in the image.")

print("=" * 60)
```

```python
# create a string that holds all detected texts separated by spaces
detected_text_string = ' '.join(texts)
print(detected_text_string)

# detect number, and exctract it along with the word next to it
import re
def extract_dosage_and_quantity(text):

    dosage_match = re.search(r"\b\d+\.?\d*\s*(?:mg|g|mcg|ml|mL|IU|%)\b", text, re.IGNORECASE)
    dosage = dosage_match.group(0).replace("mg", " mg") if dosage_match else None

    qty_match = re.search(r"\b\d+\s*(?:tablets?|capsules?|tabs?|caps?|vials?)\b", text, re.IGNORECASE)
    quantity = qty_match.group(0) if qty_match else None

    return {"dosage": dosage, "quantity": quantity}

print (extract_dosage_and_quantity(detected_text_string))
```
